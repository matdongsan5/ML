```python
## 모듈로딩
import torch
import torch.nn as nn

from torchvision.models import resnet18, resnet50, ResNet18_Weights, ResNet50_Weights
```

```python
##모델별 인스턴스 생성
model18 = resnet18(weights=ResNet18_Weights.DEFAULT)
model18
```

```python
model50 = resnet50(weights = ResNet50_Weights.DEFAULT)
model50
```

```python
##모델층별파라미터
for name, param in model18.named_parameters():
    print(name, param.shape)
```

```python
##모델층별파라미터 - 이름
for name, param in model18.named_children():
    print(name)
```
