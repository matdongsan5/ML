```python
# %pip install tqdm
import os
import zipfile
from tqdm import tqdm
```

```python
dataPath = './_data/open.zip'
extract_to= './_data/'
```

```python
file_size = os.path.getsize(dataPath)
print(f"파일 크기: {file_size / 1024:.2f} KB")
print(f"파일 크기: {file_size / (1024 ** 2):.2f} MB")
print(f"파일 크기: {file_size / (1024 ** 3):.2f} GB")
# 10.4 GB
```

```python

with zipfile.ZipFile(dataPath, 'r') as zip_ref:
    file_list = zip_ref.infolist()
    
    for file in tqdm(file_list, desc="Extracting", unit="file"):
        zip_ref.extract(file, extract_to)
```
