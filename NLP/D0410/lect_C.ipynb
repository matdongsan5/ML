#### 변수
- 일반 변수 : 데이터 저장 변수
                ex) abcd = "ABC"  ==>  "ABC"가 저장됨
- 포인터 변수 : 데이터 주소 저장 변수
                ex) *abcd = "ABC"  ==>  "ABC"가 저장된 주소 위치를 저장.

typedef struct _USERDATA{
     int nAge;
