```python
import pandas as pd
```

```python
dataDF = pd.read_csv('./dataDF.csv',index_col=0,  encoding='utf-8')
```

```python
Dialogue,age = [], []
with open('./_data/1987.txt', 'r',encoding='utf-8') as f:
    data = f.readlines()
    for line in data[2:]:
        Dialogue.append(line.replace('\n',''))
        age.append(80)
```

```python
df1987 = pd.DataFrame(columns=['Dialogue','age'])
df1987['Dialogue'] = Dialogue
df1987['age'] = age
df1987.to_csv('./1987.csv', encoding='utf-8')
```

```python
dataDF=pd.concat((dataDF ,df1987)).reset_index(drop=True)
```
