#### [ 자연어 전처리 - 토큰화 ]
- 의미있는 단위로 나누는 작업 ==> 토큰화

[1] 모듈 로딩 <hr>

```python
## NLTK 패키지의 토큰화 모듈 로딩
from nltk.tokenize import * 
import string 
```

[2] 단어 단위 토큰화 <hr>

```python
## 문장 데이터
txt = "Happy, New Year! Don't stop. Dh.Park!"
```

```python
## 단어 단위 토큰화  : 약어, 축약어, 문장부호 등을 좀 더 자연스럽게 처리
result = word_tokenize(txt)
```

```python
## 결과 출력
print(f'result \n{result}')
```

[3] 구두점 제거 후 단어 토큰화<hr>

```python
## 구두점에 해당 문자
PUNCT=string.punctuation
```

```python
## 1차 정제 ==> 문장에서 구두점 제거
for ch in txt:
    if ch in PUNCT:
        txt=txt.replace(ch, '')
txt
```

```python
## 토큰화 진행
result = word_tokenize(txt)
print(result)
```

```python
## => 2차 정제 : 불필요한 토큰 제거
## - 소문자로 일치
txt=txt.lower()
print(txt)

## - 불필요 토큰 dont 제거
txt=txt.replace('dont','')
print(txt)

## - 불필요 토큰 dhpark 제거
txt=txt.replace('dhpark','')
print(txt)
```

```python
## 토큰화 진행
result = word_tokenize(txt)
print(result)
```
