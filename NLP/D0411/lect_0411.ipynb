[일반 ANN]

- 데이터 파일 => 그대로
                 * 피쳐 : 스케일링
                 * 타겟: 정수화

* 딥러닝의 대량 데이터 처리 위해서 PyTorch에서 사용하는 기법 ---
- CustomDataset ==> 딥러닝의 대량 데이터 처리 위해서 사용
                    인덱스에 해당하는 샘플 => 피처와 타겟을 묶음으로 변환
                    사용자의 데이터에 맞도록 전처리하고 텐서 변환

- DataLoader ==> 학습에 사용할 만큼만 추출 => 텐서 타입의 피쳐와 타겟


[이미지 CNN]
- 이미지 파일 => 이미지 데이터 읽기 => 파일에 저장.(숫자값)
                 * 피쳐 : 스케일링(0~255) => 0.0~ 1.0
                 * 타겟: 정수화

* 딥러닝의 대량 데이터 처리 위해서 PyTorch에서 사용하는 기법 ---
- CustomDataset ==> 딥러닝의 대량 데이터 처리 위해서 사용
                    인덱스에 해당하는 샘플 => 피처와 타겟을 묶음으로 변환
                    사용자의 데이터에 맞도록 전처리하고 텐서 변환

- ImageFolder => 이미지 파일이 들어있는 폴더 지정하면 데이터셋 생성.(경로만있음)
                    인덱스에 해당하는 이미지를 로딩해서 전처리, 
                    rawdata, lanbel 텐서로 반환

- DataLoader ==> 학습에 사용할 만큼만 추출 => 텐서 타입의 피쳐와 타겟


[텍스트 RNN]
- 텍스트 파일 => 글자
                => 텍스트 전용 단어사전/집합 VOCAB 
                * VOCAB 만들기
                    - 토큰화 => 정수 숫자 부여
                - 피처 : 단어사전/집합기반으로 -> 숫자
                - 타겟 : 
                    - 문장/단어의 경우 단어사전/집합기반으로 -> 숫자
                    - 범주형 -> 숫자

- 문장의 길이가 다름 => 길이 맞추기.: 패딩

- 텍스트에 피쳐를 줄이기위해서 => 임베딩: 고차원 ==> 저차원: 단어정보 손실. ==> 임베딩모델
                                Embedding Layer: 학습 통해서
                                사전에 대량의 데이터를 학습한 임베딩 모델.:gensim

* 딥러닝의 대량 데이터 처리 위해서 PyTorch에서 사용하는 기법 ---
- CustomDataset ==> 딥러닝의 대량 데이터 처리 위해서 사용
                    인덱스에 해당하는 샘플 => 피처와 타겟을 묶음으로 변환
                    사용자의 데이터에 맞도록 전처리하고 텐서 변환

- DataLoader ==> 학습에 사용할 만큼만 추출 => 텐서 타입의 피쳐와 타겟
