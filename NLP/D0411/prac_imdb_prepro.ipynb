```python
# [3] 데이터로더 생성 <hr>
import torch
import torch.nn as nn
import numpy as np
import pandas as pd

## 데이터셋, 데이터로더 관련 모듈
from torch.utils.data import DataLoader, TensorDataset
from torch.nn.utils.rnn import pad_sequence
## 데이터 길이 맞추기

## 토커나이저, 단어사전 관련 모듈
from torchtext.data.utils import get_tokenizer              ## 토커나이저 인스턴스 추출
from torchtext.vocab import build_vocab_from_iterator       ## 데이터셋에서 단어사전 생성 함수
from nltk.corpus import stopwords                           ## 불용어 데이터셋
import nltk
```

```python
data_dir='data/'
data_file=data_dir+'IMDb_Reviews.csv'
```

```python
# DS만들기
```

```python
import string
PUNC = string.punctuation
NUMS = [str(x) for x in range(10)]
```

```python
### ==> 특별 문자 토큰
UNK, PAD  = '<UNK>',  '<PAD>'
STOPWORDS = stopwords.words('english')
print(f'STOPWORDS : {STOPWORDS[:10]}')

### ==> 토커나이즈 생성
tokenizer = get_tokenizer('basic_english')
```

```python
result = []
with open(data_file, 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()  # 줄바꿈 제거
        line = ''.join([char for char in line if char not in PUNC])  # 문장부호 제거
        line = ''.join([char for char in line if char not in NUMS])  # 문장부호 제거
        words = tokenizer(line)  # 단어 분리
        result.append(words)
```
