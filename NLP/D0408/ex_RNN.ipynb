```python
import torch
import torch.nn as nn
```

```python
## RNN 모델 관련 설정값
input_size = 128
output_size = 256
num_layer = 3
bid = True
BATCH_SIZE = 4
SEQUENCE_LEND = 6
```

```python
## 임의의 데이터
datas =torch.randn(BATCH_SIZE, SEQUENCE_LEND, input_size, output_size)
datas.shape
```

```python
model = nn.RNN(
    input_size=input_size,
    hidden_size= output_size,
    num_layers=num_layer,
    nonlinearity='tanh',
    batch_first=True,
    bidirectional=bid
)
```

```python
inputs = torch.rand(BATCH_SIZE, SEQUENCE_LEND, input_size)
h_0 = torch.randn(num_layer* (int(bid)+1), BATCH_SIZE, output_size)
```

```python
outputs, hidden = model(inputs, h_0)
print(outputs.shape)
print(hidden.shape)
```
