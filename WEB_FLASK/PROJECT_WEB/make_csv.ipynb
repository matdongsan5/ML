```python
import os
import pandas as pd
```

```python
trainlabel = os.listdir('./data/archive/training')
trainlist = []
for i in trainlabel:
    for j in os.listdir('./data/archive/training/'+i):
        trainlist.append(['./data/training/'+i+'/'+j, i])
        
trainDF = pd.DataFrame(trainlist, columns=['path','label'])
trainDF.head()
trainDF.to_csv('trainlist.csv', index=False)
```

```python
validlabel = os.listdir('./data/archive/validation')
validlist = []
for i in validlabel:
    for j in os.listdir('./data/archive/validation/'+i):
        validlist.append(['./data/validation/'+i+'/'+j, i])
        
validDF = pd.DataFrame(validlist, columns=['path','label'])
validDF.head()
validDF.to_csv('validlist.csv', index=False)
```

```python
testlabel = os.listdir('./data/archive/test')
testlist = []
for i in testlabel:
    for j in os.listdir('./data/archive/test/'+i):
        testlist.append(['./data/test/'+i+'/'+j, i])
        
testDF = pd.DataFrame(testlist, columns=['path','label'])
testDF.head()
testDF.to_csv('testlist.csv', index=False)
```
