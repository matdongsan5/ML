```python
# %pip install pandas sqlalchemy mysql-connector-python
import mysql.connector
import pandas as pd
```

```python
# MySQL 데이터베이스 연결

## ***** 전체 데이터(다운로드 받은대로)
##          ++ 경로수정까지
## 수정할 부분
##  user, password 자기걸로
##  query_n0_n1  ==> 자기담당으로
##  query_n2_n9  --> 자기 담당 아닌걸로
##  
##  table명 수정


connection = mysql.connector.connect(
    host='192.168.2.170',
    # host='172.30.1.80',
    user='jw',
    password='1234',
    database='monkeydb'
)

# SQL 쿼리 작성
query_n0_n1 = "SELECT path, label FROM trainbase WHERE label IN ('n0', 'n1');"

query_n2_n9 = """
(SELECT path, label FROM trainbase WHERE label = 'n2' LIMIT 16)
UNION ALL
(SELECT path, label FROM trainbase WHERE label = 'n3' LIMIT 16)
UNION ALL
(SELECT path, label FROM trainbase WHERE label = 'n4' LIMIT 16)
UNION ALL
(SELECT path, label FROM trainbase WHERE label = 'n5' LIMIT 16)
UNION ALL
(SELECT path, label FROM trainbase WHERE label = 'n6' LIMIT 16)
UNION ALL
(SELECT path, label FROM trainbase WHERE label = 'n7' LIMIT 16)
UNION ALL
(SELECT path, label FROM trainbase WHERE label = 'n8' LIMIT 16)
UNION ALL
(SELECT path, label FROM trainbase WHERE label = 'n9' LIMIT 16);
"""

# 데이터 로드
n0_n1_df = pd.read_sql(query_n0_n1, connection)
n2_n9_df = pd.read_sql(query_n2_n9, connection)

# 'etc' 라벨을 적용
n2_n9_df['label'] = 'etc'

# 두 데이터프레임 합치기
final_df = pd.concat([n0_n1_df, n2_n9_df], ignore_index=True)

# 새로운 테이블로 저장
cursor = connection.cursor()

# 기존에 테이블이 있으면 삭제하고 새로 생성
cursor.execute("DROP TABLE IF EXISTS n0n1_train;")
cursor.execute("""
    CREATE TABLE n0n1_train (
        id INT AUTO_INCREMENT NOT NULL,
        path VARCHAR(100),
        label VARCHAR(30),
        PRIMARY KEY(id)
    );
""")

# 데이터프레임을 새로운 테이블에 삽입
for row in final_df.itertuples():
    cursor.execute("""
        INSERT INTO n0n1_train (path, label)
        VALUES (%s, %s);
    """, (row.path, row.label))

connection.commit()
cursor.close()
connection.close()

print("새로운 테이블이 생성되었습니다.")
```
