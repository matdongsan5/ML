 - 학습 : 지도학습 + 회귀
 - 알고리즘 : 인공신경망
 

```python
""" 
모델 클래스 설계
클래스이름 Env_predict
부모 클래스 nn.Module
모델층구성      입력 신호수     출력신호수      활성화함수
입력층           3                50               Relu
                         ↙️
은닉층1          50               30                 Relu    
                         ↙️
은닉층2          30               10               Relu
                         ↙️
출력층           10                 1               

"""
```

```python
featureDF = pd.DataFrame({0: [1,2,3], 1:[4,5,6], 2:[7,8,9]})
targetSR = pd.Series([0,1,2])
```

```python

class Env_predict(nn.Module):
    ## 모델 층 구성 관련 초기화
    def __init__(self):
        super().__init__()
        
        self.in_layer = nn.Linear(3, 50)
        self.hd_layer1 = nn.Linear(50, 30)
        self.hd_layer2 = nn.Linear(30, 10)
        self.out_layer = nn.Linear(10, 1)
        ## 선형층 인스턴스 만들어줌.
    
    ## 학습 즉, 순방향 학습 진행 메서드 foward : 필수 오버라이딩 메서드
    ## 매개변수 data : 학습용 데이터 전달
    def forward(self, data):
        # 입력 ==> 입력층 학습
        out =  self.in_layer(data)     ## 출력: (f1*w11 + f2*w12 + f3*w13 + b1) + ... + (f1*w501 + f2*w502 + f3*w503 + b50)
        out = F.relu(out)
        
        # 은닉층
        out =  self.hd_layer1(out)     ## 출력: (f1*w11 + f2*w12 + f3*w13 + .... f50*w150 + b1) + ... + (f1*w301 + f2*w302 + f3*w303 + ... + f50*w3050 + b30)
        out = F.relu(out)
        
        # 은닉층
        out =  self.hd_layer2(out)     ## 출력: (f1*w11 + f2*w12 + f3*w13 + .... f30*w130 + b1) + ... + (f1*w101 + f2*w102 + f3*w103 + ... + f30*w1030 + b10)
        out = F.relu(out)
        
        # 출력층
        out = self.out_layer(out)       ## 출력 f1*w1 +..... + f10*w10 + b1
        
        return out
        
```

```python
## 모델 설계 체크

model = Env_predict()
model(torch.FloatTensor([[22,33,44]]) ) # 1,3 
```

문제 2 이미지 분류 모델
 - 이미지 크기 30*30 흑백 
 - 피쳐 900 (30*30*1)
 - 타겟 자동차/나무
 - 학습: 지도학습 - 분류 ==> 이진분류

```python
""" 
모델 클래스 설계
클래스이름 Env_predict
부모 클래스 nn.Module
모델층구성      입력 신호수     출력신호수      활성화함수
입력층           900             250               Relu
                         ↙️
은닉층1          250             150                 Relu    
                         ↙️
은닉층2          150               70               Relu
                         ↙️
출력층           70                 1               sigmoid

"""
```

```python

class Env_predict(nn.Module):
    ## 모델 층 구성 관련 초기화
    def __init__(self):
        super().__init__()
       
        self.in_layer = nn.Linear(900, 250)
        self.hd_layer1 = nn.Linear(250, 150)
        self.hd_layer2 = nn.Linear(250, 130)
        self.hd_layer3 = nn.Linear(130, 70)
        self.out_layer = nn.Linear(70, 1)
        ## 선형층 인스턴스 만들어줌.
    
    ## 학습 즉, 순방향 학습 진행 메서드 foward : 필수 오버라이딩 메서드
    ## 매개변수 data : 학습용 데이터 전달
    def forward(self, data):
        # 입력 ==> 입력층 학습
        out =  self.in_layer(data)     ## 출력: (f1*w11 + f2*w12 + f3*w13 + ... + f900*w1900+ b1) + ... + (f1*w9001 + f2*w9002 + f3*w9003 + ... + f900*w1900 + b900)
        out = F.relu(out)
        
        # 은닉층
        out =  self.hd_layer1(out)     ## 출력: (f1*w11 + f2*w12 + f3*w13 + .... f50*w150 + b1) + ... + (f1*w1501 + f2*w1502 + f3*w1503 + ... + f50*w15050 + b150)
        out = F.relu(out)
        
        # 은닉층
        out =  self.hd_layer2(out)     ## 출력: (f1*w11 + f2*w12 + f3*w13 + .... f30*w130 + b1) + ... + (f1*w101 + f2*w102 + f3*w103 + ... + f30*w1030 + b10)
        out = F.relu(out)
        
        # 은닉층
        out =  self.hd_layer3(out)     ## 출력: (f1*w11 + f2*w12 + f3*w13 + .... f30*w130 + b1) + ... + (f1*w101 + f2*w102 + f3*w103 + ... + f30*w1030 + b10)
        out = F.relu(out)
        
        # 출력층
        out = self.out_layer(out)       ## 출력 f1*w1 +..... + f10*w10 + b1
        out = F.sigmoid(out)
        return out
        
```

```python

class Env_predict(nn.Module):
    ## 모델 층 구성 관련 초기화
    def __init__(self):
        super().__init__()
        self.layers = nn.Sequential(
                        nn.Linear(900, 250),
                        nn.ReLU(),
                        nn.Linear(250, 150),
                        nn.ReLU(),
                        nn.Linear(250, 130),
                        nn.ReLU(),
                        nn.Linear(130, 70),
                        nn.ReLU(),
                        nn.Linear(70, 1),
                        nn.Sigmoid()
        )
        ## nn.Sequential() 사용. 순서대로 적용.
        ## 선형층 인스턴스 만들어줌.
    
    ## 학습 즉, 순방향 학습 진행 메서드 foward : 필수 오버라이딩 메서드
    ## 매개변수 data : 학습용 데이터 전달
    def forward(self, data):
        return self.layers(data)
```
