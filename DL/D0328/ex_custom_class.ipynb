#### [ 커스텀 클래스의 오버라이딩 메서드 ]


```python
class Test:
    # 클래스 초기화 기능의 메서드 인스턴스(속성) 메모리 저장 기능
    # 생성자(Constructor): -- Test()
    # 클래스명 호출시. __init__가 실행되도록 매핑되어 있음.
    def __init__(self, num, name):
        self.num = num
        self.name = name
    
    def __str__(self):
        print("__str__()")
        
    def __add__(self, other):
        print("__add__()")
        return self.num+ other.num
     
    def __mul__(self, other):
        print("__mul__()")   
        return self.num ** other.num
    
    def __call__(self, data):
        print('__call__()')
        return "callign'"
```

```python
t1 = Test(10, 'A')
t2 = Test(5, 'B')
```

```python
t1*t2
```
