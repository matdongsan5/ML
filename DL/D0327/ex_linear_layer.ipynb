```python
import torch               
#텐서 관련 모듈

import torch.nn as nn
# 신경망 관련 모듈

import torch.nn.functional as F
# 활성화 및 신경망 관련 함수들
```

```python
#임시데이터
X_train  = torch.randn(2,4)
y_train  = torch.randint(1,5,(2,1))
```

```python
print( X_train.shape, y_train.shape)
```

[2] 선형결합층(Full-Connected Layer : FC Layer)

```python
## 인스턴스 생성
layer = nn.Linear(4,3)     # 입력 피쳐 수, 출력 결과 수 =  퍼셉트론 수

#
```

```python
## 가중치 합 계산
output = layer(X_train)
output
```

```python
## W, b 텐서 확인
# 퍼셉트론 개수만큼 결과를 뱉어냄.
print(f"weight => {layer.weight}")
print(f"bias => {layer.bias}")
```
