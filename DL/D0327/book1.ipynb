```python
import torch

# 모델의 가중치 (랜덤 초기화)
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

# 옵티마이저 설정 (SGD)
optimizer = torch.optim.SGD([w, b], lr=0.01)

# 학습 반복
for i in range(100):
    loss = (w - 2) ** 2  # 손실 함수 (간단한 예제)

    optimizer.zero_grad()  # 기울기 초기화
    loss.backward()  # 기울기 계산
    optimizer.step()  # 가중치 업데이트

    if i % 10 == 0:
        print(f"Iteration {i}: Loss = {loss.item()}, w = {w.item()}")
```



4. 옵티마이저 선택 가이드
💡 어떤 옵티마이저를 선택해야 할까?

✅ SGD

데이터가 많고 연산량을 줄이고 싶다면 사용

하지만 일반적으로 Momentum을 함께 사용하는 것이 좋음

✅ Adam

대부분의 경우 기본적으로 Adam을 사용하면 안정적

특히 CNN, RNN, Transformer 등 다양한 모델에서 활용됨

✅ RMSprop

CNN(합성곱 신경망) 모델에서 좋은 성능을 보임

특히 이미지 처리(컴퓨터 비전)에서 많이 사용됨

✅ Adagrad

자주 업데이트되는 가중치는 학습률을 줄이기 때문에 NLP(자연어 처리)에서 자주 사용됨
