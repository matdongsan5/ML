#### [ Tensor & Numpy 집계함수]

```python
import numpy as np
import torch
```

by numpy

```python
data = np.array([[11,22,33], [44,11,55]])
data
```

```python
print(data.max())
print(data.max(axis=0)) ## 세로축 중에서 젤 높은것.
print(data.max(axis=1))## 가로 축 중에서 . 
```

```python
## ndarray에서 큰 값의 인덱스 찾기.
print(data.argmax())
print(data.argmax(axis=0))
print(data.argmax(axis=1))
```
