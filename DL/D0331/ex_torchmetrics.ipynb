#### [Torchmetrics 모듈 활용]

```python
from torchmetrics.classification import *  #분류
from torchmetrics.regression import *   #회귀

import torch
```

```python
##분류용 성능지표 함수들
target = torch.tensor([0,1,0,1])
preds = torch.tensor([0,1,1,0])
preds2 = torch.tensor([.11, 0.56, 0.76, 0.51])

accuracy = BinaryAccuracy()
```

```python
accuracy(preds,target)
```

```python
accuracy(preds2,target) ## threshold 0.5 기준으로 0과 1로 결과 추출
```

```python
## 다중분류
target = torch.tensor([2, 1, 0, 0])
preds = torch.tensor([2, 1, 0, 1])

metric = MulticlassAccuracy(num_classes=3)
metric(preds, target)
```

```python
mca = MulticlassAccuracy(num_classes=3, average=None)
mca(preds, target)
```

```python
target = torch.tensor([2, 1, 0, 0])
preds = torch.tensor([[0.16, 0.26, 0.58],
                [0.22, 0.61, 0.17],
                [0.71, 0.09, 0.20],
                [0.05, 0.82, 0.13]])
metric = MulticlassAccuracy(num_classes=3)
print('acc', metric(preds, target))

f1 = MulticlassF1Score(num_classes=3)
print('f1', f1(preds, target))

prescision = MulticlassPrecision(num_classes=3)
print('Precision', prescision(preds, target))

recall = MulticlassRecall(num_classes=3)
print('recall', recall(preds, target))
```

```python

mca = MulticlassAccuracy(num_classes=3, average=None)
mca(preds, target)
```

```python
# 회귀용 성능지표 : mse, mae, rmse, r2


target = torch.tensor([3.0, -0.5, 2.0, 7.0])
preds = torch.tensor([2.5, 0.0, 2.0, 8.0])

# MAE
mean_absolute_error = MeanAbsoluteError()
print('mae', mean_absolute_error(preds, target))

# MsE
mse = MeanSquaredError()
print('mse', mse(preds, target))

# r2
r2 = R2Score()
print('r2', r2(preds, target))
```
