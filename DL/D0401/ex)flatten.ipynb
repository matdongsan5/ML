### flatten

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchmetrics
import torchinfo
```

```python
# 입력 데이터 생성
t = torch.tensor([[[ 1,2 ], [3,4], ] , [[5,6], [7,.8]]])
```

```python
print(f'형 태 : {t.shape}')
print(f'차 원 : {t.ndim}')
print(f'타 입 : {t.dtype}')
```

```python
# 기본값 start_dim=1, end_dim= -1
# 입력 데이터의 shape ( d, h, w )  ( 0, 1, 2)
f=nn.Flatten()
output = f(t)
print( type(f) )
print( output.shape )
print( output )
```

```python
# 차원 변경
fLayer = nn.Flatten(start_dim=0)
outTS = fLayer(t)

print(f"형 태 : {outTS.shape}, {outTS}")
```
